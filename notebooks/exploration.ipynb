```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le dataset
df = pd.read_csv('../data/ipynb_checkpoints/Mental Health Classification.csv')
df.columns = df.columns.str.strip().str.replace('\xa0',' ', regex=True)
[col for col in df.columns if "depress" in col.lower()]

df.head()
```

```python
# Dimensions du dataset
print("Dimensions :", df.shape)
df.info()
```

```python
# Statistiques descriptives
df.describe(include='all')
```

```python
print(df.columns.tolist())

plt.figure(figsize=(12,6))
sns.heatmap(df.select_dtypes(include=['int64','float64']).corr(), annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure()
df['Depression_Type'].value_counts().plot(kind='bar')
plt.title('Distribution du type de dépression')
plt.xlabel('Type')
plt.ylabel('Nombre')
plt.show()
```
